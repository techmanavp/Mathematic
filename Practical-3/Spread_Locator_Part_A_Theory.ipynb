## 1. What is a Statistical Distribution?

**Definition:** A statistical distribution describes all the possible values a variable can take and how often (or how likely) each value occurs. It shows the *shape* and *spread* of the data.

**Key points**
- It is described by parameters such as the mean (centre) and standard deviation (spread).
- It can be shown with a histogram (for data) or with a mathematical function (for a model).
- Knowing the distribution lets us calculate probabilities, make predictions and choose the correct statistical method.

**Common distributions:** Bernoulli, Binomial, Poisson, Normal, Log-Normal, Power Law (Pareto), Exponential.

**Example:** Heights of students are usually bell-shaped (Normal). Daily orders on a website often follow a Poisson distribution. Order amounts are often right-skewed (Log-Normal).

**In this project:** We checked which distribution best describes transaction success, weekly counts, daily counts and transaction amounts.

## 2. What is a Q-Q Plot and why is it used?

**Definition:** A Q-Q (Quantile-Quantile) plot is a graph that compares the quantiles of our data with the quantiles of a theoretical distribution (usually Normal).

**How to read it**
| What you see | Meaning |
|---|---|
| Points lie on the straight line | Data follows the chosen distribution |
| Points curve upward at the right end | Right-skewed data (long right tail) |
| S-shaped curve | Heavy tails (more extreme values than normal) |
| Points far from line at the ends only | Outliers |

**Why it is used**
- To visually test normality before using methods that assume normal data.
- To compare the fit of a distribution with real data.
- To check whether a transformation (log, Box-Cox) worked.

**In this project:** The Q-Q plot of transaction amount curved away from the line (not normal), but after taking the log it followed the line closely, which supports the Log-Normal model.

## 3. Difference between Discrete and Continuous Distributions

| Point | Discrete Distribution | Continuous Distribution |
|---|---|---|
| Values | Countable (0, 1, 2, 3 ...) | Any value in a range (e.g. 1250.75) |
| Function used | PMF (Probability Mass Function) | PDF (Probability Density Function) |
| Probability of an exact value | Can be greater than 0 | Always 0; we use intervals instead |
| Total probability | Sum of all PMF values = 1 | Area under the PDF curve = 1 |
| Graph | Bars | Smooth curve |
| Examples | Bernoulli, Binomial, Poisson | Normal, Log-Normal, Exponential, Power Law |
| Project example | Number of transactions per day | Transaction amount in Rs |

## 4. What is Bernoulli Distribution?

**Definition:** A Bernoulli distribution models a single trial with only two outcomes: success (1) or failure (0).

**Example:** Tossing a coin once (head = success, p = 0.5). Whether a customer's transaction is successful or failed.

**In this project:** Transaction status (Success / Fail) is Bernoulli. We found p(Success) = 0.4455 and p(Fail) = 0.5545, so more than half the transactions fail.

---
## 5. What is Binomial Distribution?

**Definition:** A Binomial distribution gives the number of successes in *n* independent Bernoulli trials, each with the same success probability *p*. (It is the sum of n Bernoulli trials.)


**Conditions:** fixed number of trials n, only two outcomes, independent trials, constant p.

**Example:** Tossing a fair coin 10 times, the probability of exactly 6 heads is $\binom{10}{6}(0.5)^{10} = 210/1024 \approx 0.205$.

**In this project:** Weekly transaction count per customer was modelled as Binomial with n = 9 (maximum weekly count, assumed because no fixed n is given) and p = 0.317. Mean matched (2.85) but actual variance (3.23) was higher than the Binomial variance (1.95).

## 6. Explain Log-Normal Distribution

**Definition:** A variable X is Log-Normal if its logarithm, ln(X), follows a Normal distribution. It is defined only for positive values and is right-skewed.

**Why it occurs:** When a value is the *product* of many small independent positive factors (just like a Normal comes from a sum), the result is Log-Normal.

**Examples:** Income, house prices, stock prices, customer spending, time to complete a task.

**In this project:** Transaction amounts are right-skewed with a few very large purchases. The Log-Normal fit had KS p-value = 0.90 and the lowest AIC, so it was the best fit.

---
## 7. Explain Power Law Distribution

**Definition:** In a Power Law distribution, the probability of a value decreases as a power of the value. Small values are very common and very large values are rare but possible (a heavy tail).

- $\alpha$ is the exponent (smaller alpha means a heavier tail).
- It is scale-free: it looks similar on every scale. On a log-log plot it appears as a straight line.

**Examples:** Wealth distribution (the 80/20 rule), city populations, website traffic, word frequencies.

**Log-Normal vs Power Law:** Both have long right tails, but a Power Law has a much heavier tail and its highest density is at the minimum value, whereas a Log-Normal rises first and then falls.

**In this project:** The Power Law (Pareto) fit was rejected (KS p-value close to 0) because most transactions are moderate in size, not concentrated at the minimum.

## 8. What is Box-Cox Transform?

**Definition:** Box-Cox is a power transformation that converts skewed, positive data into data that is closer to normal and stabilises its variance.

**Common lambda values**
| Lambda | Transform |
|---|---|
| 1 | No change |
| 0.5 | Square root |
| 0 | Natural log |
| -1 | Reciprocal |

**Key points**
- Works only for positive data (x > 0).
- The best lambda is found automatically by maximum likelihood (`scipy.stats.boxcox`).
- Useful before applying models that assume normality and constant variance (e.g. linear regression, t-tests).

**In this project:** Best lambda = -0.18 (close to 0, similar to log). Skewness dropped from 3.73 to about -0.01 and the Shapiro p-value became 0.64, so the transformed data is normal.

**Conditions:** events are independent, occur at a constant average rate, and cannot happen at exactly the same instant.

## 10. What is Z-score Probability?

**Definition:** The Z-score tells how many standard deviations a value is away from the mean. Z-score probability is the probability found for that Z-score using the standard normal distribution (mean 0, SD 1).


**Empirical rule (Normal data):** about 68% of values lie within 1 SD, 95% within 2 SD and 99.7% within 3 SD. A value with |Z| > 3 is usually treated as an outlier.

**Example:** For exam marks with mean 60 and SD 10, a student with 75 marks has Z = (75 - 60)/10 = 1.5. $P(X > 75) = 1 - \Phi(1.5) \approx 0.067$, so about 6.7% of students score higher.

**In this project:** For Rs 5000, Z = 0.82, which gives P(amount > 5000) of about 20.5% under the normal assumption. The actual value in the data is 11.4%, so the normal assumption overestimates because the data is skewed.

## 11. Differentiate PDF and CDF

| Point | PDF (Probability Density Function) | CDF (Cumulative Distribution Function) |
|---|---|---|
| Meaning | Shows how dense the probability is at each value | Shows probability that X is less than or equal to x |
| Graph shape | Bell, hump or other curve | Always rising S-curve from 0 to 1 |
| Value range | Can be greater than 1 (it is a density, not a probability) | Always between 0 and 1 |
| Probability of an interval | Area under the curve between a and b | F(b) - F(a) |
| Total | Total area = 1 | Ends at 1 |
| Used for | Seeing where values are concentrated and the shape of data | Finding percentiles and probabilities like P(X > 5000) |

**For discrete variables** the PDF is replaced by the PMF.

**In this project:** The PDF of transaction amount showed most transactions between Rs 2,000 and Rs 4,500 with a long right tail. The CDF showed that about 86% of transactions are Rs 5000 or less.
